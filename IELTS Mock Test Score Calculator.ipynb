```python
# IELTS模考成绩分析器
# 功能：记录模考日期、计算雅思总分并比较成绩

print("IELTS模考成绩分析器")

# 输入模考日期
date = input("请输入模考日期：")

# 输入四科成绩
listening = float(input("请输入听力成绩："))
reading = float(input("请输入阅读成绩："))
writing = float(input("请输入写作成绩："))
speaking = float(input("请输入口语成绩："))

# 计算总分
average = (listening + reading + writing + speaking) / 4
overall = int(average * 2 + 0.5) / 2

# 输出本次模考成绩
print("\n本次模考成绩")
print("日期：", date)
print("听力：", listening)
print("阅读：", reading)
print("写作：", writing)
print("口语：", speaking)
print("雅思总分：", overall)

# 与上一次模考成绩比较
last_score = input("\n请输入上次模考总分（首次可直接回车）：")

if last_score != "":
    change = overall - float(last_score)
    print("与上次相比：", change)

    if change > 0:
        print("成绩提高啦！")
    elif change < 0:
        print("成绩下降咯，继续努力~")
    else:
        print("成绩与上次持平呢")
```
